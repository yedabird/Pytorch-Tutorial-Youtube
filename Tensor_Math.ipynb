```python
import torch
import numpy as np
```

```python
# add
tensor_1 = torch.tensor([1,2,3,4,5])
tensor_2 = torch.tensor([6,7,8,9,10])
print(f"tensor-1: {tensor_1}")
print(f"tensor-2: {tensor_2}")
tensor_3 = tensor_1 + tensor_2
print(f"tensor-3: {tensor_3}")
print(f"tensor_1.add(tensor_2): {tensor_1.add(tensor_2)}")
```

```python
# substraction
tensor_1_minus_2 = tensor_1 - tensor_2
print(f"tensor_1_minus_2: {tensor_1_minus_2}")
print(f"tensor_1.sub(tensor_2): {tensor_1.sub(tensor_2)}")
#
```

```python
# mulplication
tensor_1_mul_2 = tensor_1 * tensor_2
print(f"tensor_1_mul_2: {tensor_1_mul_2}")
print(f"tensor_1.mul(tensor_2): {tensor_1.mul(tensor_2)}")
#
```

```python
# division
tensor_1_div_2 = tensor_1 / tensor_2
print(f"tensor_1_div_2: {tensor_1_div_2}")
print(f"tensor_1.div(tensor_2): {tensor_1.div(tensor_2)}")
#
```

```python
# modular
print(f"tensor_1: {tensor_1}")
print(f"tensor_2: {tensor_2}")
tensor_1_mod_2 = tensor_1 % tensor_2
print(f"tensor_1_mod_2: {tensor_1_mod_2}")
tensor_2_mod_1 = tensor_2 % tensor_1
print(f"tensor_2_mod_1: {tensor_2_mod_1}")

#
```

```python
# remainder
print(f"tensor_1: {tensor_1}")
print(f"tensor_2: {tensor_2}")
tensor_1_remainder_2 = torch.remainder(tensor_1, tensor_2)
print(f"tensor_1_remainder_2: {tensor_1_remainder_2}")
tensor_2_remainder_1 = torch.remainder(tensor_2, tensor_1)
print(f"tensor_2_remainder_1: {tensor_2_remainder_1}")
#
#
```

```python
# exponent
print(f"tensor_1: {tensor_1}")
print(f"tensor_2: {tensor_2}")
tensor_1_exp_2 = tensor_1 ** tensor_2
print(f"tensor_1_exp_2: {tensor_1_exp_2}")
tensor_2_exp_1 = tensor_2 ** tensor_1
print(f"tensor_2_exp_1: {tensor_2_exp_1}")
#
```

```python
tensor_x = torch.tensor([1,2,3,4,5])
tensor_y = torch.tensor([6,7,8,9,10])
print(f"tensor_x: {tensor_x}")
print(f"tensor_y: {tensor_y}")
```

```python
tensor_x.add(tensor_y)
print(f"tensor_x using add: {tensor_x}")
```

```python
tensor_x.add_(tensor_y)
print(f"tensor_x using add_: {tensor_x}")
#
```
