```python
import torch
```

```python
my_torch = torch.arange(10)
my_torch
```

```python
# reshape
my_torch = my_torch.reshape(2,5)
my_torch
```

```python
# reshape if we don't know the number of items using -1
my_torch2 = torch.arange(10)
print(f"my_torch2: {my_torch2}")
my_torch2 = my_torch2.reshape(-1,5)
print(f"my_torch2(-1,5): {my_torch2}")
my_torch2 = my_torch2.reshape(5,-1)
print(f"my_torch2(5,-1): {my_torch2}")
```

```python
# reshape if we don't know the number of items
my_torch3 = torch.arange(15)
print(f"my_torch3: {my_torch3}")
my_torch3 = my_torch3.reshape(-1,3)
print(f"my_torch3(-1,3): {my_torch3}")
my_torch3 = my_torch3.reshape(3,-1)
print(f"my_torch3(3,-1): {my_torch3}")
```

```python
my_torch4 = torch.arange(10)
print(f"my_torch4: {my_torch4}")
my_torch4_view = my_torch4.view(2,5)
print(f"my_torch4_view: {my_torch4_view}")
```

```python
## difference between reshape and view in pytorch
# https://stackoverflow.com/questions/49643225/whats-the-difference-between-reshape-and-view-in-pytorch
```

```python
##= with reshape and view, they will update
my_torch5 = torch.arange(10)
print(f"my_torch5: {my_torch5}")
my_torch5_reshape = my_torch5.reshape(2,5)
print(f"my_torch5_reshape: {my_torch5_reshape}")
my_torch5[1] = 4141
print(f"my_torch5: {my_torch5}")
print(f"my_torch5_reshape: {my_torch5_reshape}")
```

```python
##= with reshape and view, they will update
my_torch6 = torch.arange(10)
print(f"my_torch5: {my_torch6}")
my_torch6_view= my_torch6.view(2,5)
print(f"my_torch5_view: {my_torch6_view}")
my_torch6[1] = 100
print(f"my_torch6: {my_torch6}")
print(f"my_torch6_view: {my_torch6_view}")
```

```python
## slice
my_torch7 = torch.arange(10)
print(f"my_torch7: {my_torch7}")
my_torch8 = my_torch7.view(5,2)
print(f"my_torch8: {my_torch8}")
print(f"my_torch8[:,0]: {my_torch8[:,0]}")
print(f"my_torch8[0,:]: {my_torch8[0,:]}")
```
